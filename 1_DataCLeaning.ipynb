```python
import pandas as pd 
import numpy as np
```

```python
#Initialise a dataframe to Full.csv in data folder
df = pd.read_csv('Data/Full.csv')
```

```python
# Drop all columns that contain the string "Unnamed" in their name
df = df.drop(df.filter(regex='Unnamed').columns, axis=1)
```

```python
df
```

```python
#Delete rows with all NaN values
df = df.dropna(how='all')
df
```

```python
#Updating the csv file after cleaning
df.to_csv('Data/Full_Cleaned.csv', index=False)
```

```python
#Add a column Year in dataframe store null for now
df['Year'] = np.nan
```

```python
year = 2020
for i in range(len(df)):
    if df.iloc[i]['Sl. No.'] > year:
        year = year + 1
    df.at[i, 'Year'] = year
```

```python
df
```

```python
#delete rows with Sl. No. as 2020,2021,2022,2023,2024 and 2025
df = df[~df['Sl. No.'].isin([2020, 2021, 2022, 2023, 2024, 2025])]
```

```python
#Updating the csv file after cleaning
df.to_csv('Data/Full_Cleaned.csv', index=False)
```
